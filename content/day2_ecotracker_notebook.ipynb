```python
def get_data():
    distance = float(input("How far did you walk today (km)? "))
    food = input("What did you eat? ")
    energy = float(input("How much energy did you use (kWh)? "))
    waste = float(input("Waste produced (kg)? "))

    return {
        "distance": distance,
        "food": food,
        "energy": energy,
        "waste": waste
    }
```

```python
def calculate_scores(data):
    scores = {
        "Mobility": score_transport(data["distance"]),
        "Food": score_food(data["food"]),
        "Energy": score_energy(data["energy"]),
        "Waste": score_waste(data["waste"])
    }
    return scores
```

```python
def score_energy(energy):
    if energy < 5:
        return 10
    elif energy < 10:
        return 5
    return 2
```

```python
def score_food(food):
    food = food.lower().strip()
    if food in {"vegetarian", "vegetable", "vegetables", "vegan"}:
        return 10
    elif food in {"chicken", "poultry", "meat"}:
        return 8
    elif food in {"chocolate", "fries", "fastfood"}:
        return 4
    return 1
```

```python
def score_waste(waste):
    if waste < 0.5:
        return 10
    elif waste < 1:
        return 5
    return 2
```

```python
def score_transport(distance):
    if distance > 10:
        return 10
    elif distance > 5:
        return 6
    return 2
```

```python
def category_status(score):
    if score >= 8:
        return "GOOD"
    elif score >= 5:
        return "OKAY"
    elif score >= 3:
        return "You Need to Improve"
    return "LOW"
```

```python
def show_dashboard(scores):
    total = round(sum(scores.values()) / 40 * 100)
    print('-----------------------------------')
    print(f"Your Eco Score: {total} / 100")
    print('-----------------------------------\n')

    for category, score in scores.items():
        print(f"{category}: {category_status(score)} ({score}/10)")

    lowest, tip = recommendation(scores)

    print(f"\nTip: {tip}")
    print(f"Weakest category: {lowest}")
    print("=" * 35)

    return total, lowest
```

```python
data = get_data()
scores = calculate_scores(data)
show_dashboard(scores)
```
